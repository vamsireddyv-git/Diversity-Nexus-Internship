## Simple Ideas and Results

### 1. What is a vector?
A vector is a one-dimensional list of numbers. For example, `[2, 5, 8]` is a vector.

### 2. What are vector operations?
They are calculations on vectors. Add or subtract values in the same positions, or multiply every value by one number. For example, `[2, 4] + [1, 3]` gives `[3, 7]`.

### 3. What is a matrix?
A matrix is a table of numbers arranged in rows and columns. This example has two rows and two columns:

\[ A=\begin{bmatrix}2&1\\4&3\end{bmatrix} \]

### 4. What does matrix shape mean?
Shape tells us the number of rows and columns. A matrix with 3 rows and 2 columns has shape `(3, 2)`.

### 5. What is a dot product?
Multiply matching values, then add the results. For `[2, 4]` and `[3, 1]`, the answer is `(2 × 3) + (4 × 1) = 10`.

### 6. What is matrix multiplication?
It combines rows from the first matrix with columns from the second. For example:

\[ A=\begin{bmatrix}2&1\\4&3\end{bmatrix}, B=\begin{bmatrix}1&2\\0&1\end{bmatrix} \]

\[ AB=\begin{bmatrix}2&5\\4&11\end{bmatrix} \]

### 7. What is the difference between `*` and `@`?
`*` multiplies values in matching positions. `@` does matrix multiplication. For `[2, 4] * [3, 1]`, the result is `[6, 4]`.

### 8. What is a gradient?
A gradient tells us how quickly a value changes. For `f(x) = x^2`, the gradient is `2x`. When `x` is 4, the gradient is 8.

### 9. What is gradient descent?
It changes a model value to reduce its error. If the value is 8, the learning rate is 0.2, and the gradient is 3, the new value is `8 - (0.2 × 3) = 7.4`.

### 10. Why are vectors and matrices used in AI?
They help store and work with number data in machine learning.

### 11. Why use NumPy for linear algebra?
NumPy has fast tools for calculations with vectors and matrices. For example, `np.dot(A, B)` finds the dot product of matching vectors.

### 12. How is a vector different from a matrix?
A vector is a one-dimensional list. A matrix has rows and columns.

### 13. What if matrix sizes do not match?
The number of columns in the first matrix must equal the number of rows in the second. Otherwise, matrix multiplication cannot be done.

### 14. How do gradients help in machine learning?
They show which way to adjust model values to lower prediction error.

### 15. How are dot product and matrix multiplication different?
A dot product of two vectors gives one number. Matrix multiplication gives a new matrix when the sizes match.
