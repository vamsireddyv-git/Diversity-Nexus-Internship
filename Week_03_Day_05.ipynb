```python
# find the mean number of daily steps
import statistics
steps = [2500, 3200, 3200, 4100, 5000]
print(statistics.mean(steps))
```

```python
# find the middle number of daily steps
import statistics
steps = [2500, 3200, 3200, 4100, 5000]
print(statistics.median(steps))
```

```python
# find the most common number of daily steps
import statistics
steps = [2500, 3200, 3200, 4100, 5000]
print(statistics.mode(steps))
```

```python
# find the spread of daily step counts
import statistics
steps = [2500, 3200, 3200, 4100, 5000]
print(statistics.mean(steps))
print(statistics.pvariance(steps))
```

```python
# find how far step counts usually are from the mean
import statistics
steps = [2500, 3200, 3200, 4100, 5000]
print(statistics.pstdev(steps))
```

```python
# chance of picking a green marble
green_marbles = 4
total_marbles = 9
chance = green_marbles / total_marbles
print(chance)
```

```python
import statistics

scores = [2, 4, 6, 8, 10]

print(statistics.pvariance(scores))  # variance for the full group
print(statistics.variance(scores))   # variance for a sample

print(statistics.pstdev(scores))    # standard deviation for the full group
print(statistics.stdev(scores))     # standard deviation for a sample
```

## Simple Ideas and Results

**1. What is the mean?**  
The mean is the total divided by the number of values. For 2, 4, and 8, it is `(2 + 4 + 8) / 3`, or about 4.67.

**2. What is the median?**  
The median is the middle value after sorting. For 3, 5, and 9, the median is 5.

**3. What is the mode?**  
The mode is the value seen most often. In 4, 6, 6, and 9, the mode is 6.

**4. What is variance?**  
Variance shows how spread out the values are. For 2, 4, and 6, the mean is 4 and the population variance is about 2.67.

**5. What is standard deviation?**  
It shows a typical distance from the mean. It is the square root of variance. If variance is 9, standard deviation is 3.

**6. What is probability?**  
Probability is the chance that something will happen. If 4 of 9 marbles are green, the chance of picking green is `4 / 9`.
