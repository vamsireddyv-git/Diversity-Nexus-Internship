```python
# make a small dataset
import pandas as pd
data = {'Name': ['Vamsi', 'Abhi', 'Vamsi', 'Nila'],
        'Age': [21, None, 21, 23],
        'City': ['Hyderabad', 'Mysuru', 'Hyderabad', 'hyderabad ']}
df = pd.DataFrame(data)
print(df)
```

```python
# count empty values
import pandas as pd
df = pd.DataFrame({'Age': [19, None, 24, None]})
print(df.isnull().sum())
```

```python
# fill an empty age with the middle age
import pandas as pd
df = pd.DataFrame({'Age': [19, None, 23, 27]})
df['Age'] = df['Age'].fillna(df['Age'].median())
print(df)
```

```python
# keep rows that have an age
import pandas as pd
df = pd.DataFrame({'Age': [19, None, 24]})
print(df.dropna())
```

```python
# find and remove repeated rows
import pandas as pd
df = pd.DataFrame({'Name': ['Vamsi', 'Abhi', 'Vamsi'], 'Age': [21, 22, 21]})
print('Repeated rows:', df.duplicated().sum())
print(df.drop_duplicates())
```

```python
# change text numbers into whole numbers
import pandas as pd
df = pd.DataFrame({'Age': ['19', '22', '24']})
df['Age'] = df['Age'].astype(int)
print(df.dtypes)
```

```python
# clean spaces and use the same letter style
import pandas as pd
df = pd.DataFrame({'City': [' hyderabad', 'HYDERABAD ', 'mysuru']})
df['City'] = df['City'].str.strip().str.title()
print(df)
```

```python
# find an unusually high value with the IQR method
import pandas as pd
values = pd.Series([18, 20, 21, 22, 23, 95])
q1 = values.quantile(0.25)
q3 = values.quantile(0.75)
iqr = q3 - q1
outliers = values[(values < q1 - 1.5 * iqr) | (values > q3 + 1.5 * iqr)]
print('Unusual values:', outliers.tolist())
```

```python
# make one number column for each fruit
import pandas as pd
df = pd.DataFrame({'Fruit': ['Mango', 'Apple', 'Mango']})
encoded = pd.get_dummies(df, columns=['Fruit'], dtype=int)
print(encoded)
```

```python
# give each size a number
import pandas as pd
df = pd.DataFrame({'Size': ['Medium', 'Large', 'Small']})
df['Size_Code'] = df['Size'].map({'Small': 1, 'Medium': 2, 'Large': 3})
print(df)
```

## Simple Ideas and Results

**1. What is data cleaning?**  
It means fixing data so it is ready to use.

**2. What are missing values?**  
They are empty spaces in the data. `isnull().sum()` counts them.

**3. What is filling missing data?**  
It means replacing an empty value with a useful value. Here, the empty age is replaced by the median, 23.

**4. What does `dropna()` do?**  
It removes rows that have an empty value. The rows with an age stay.

**5. What are duplicate rows?**  
They are rows that appear more than once. `drop_duplicates()` keeps one copy of Vamsi's repeated row.

**6. What is changing a data type?**  
It changes how Python stores a value. For example, the text `'19'` becomes the number `19`.

**7. What is text cleanup?**  
It makes text look the same. Extra spaces are removed and city names use title case.

**8. What is an outlier?**  
It is a value far from most of the others. The IQR check finds 95 here. Check an outlier before removing it.

**9. What is one-hot encoding?**  
It makes a separate 0-or-1 column for each category, such as Apple and Mango.

**10. What is number encoding?**  
It changes categories into numbers. Here, Small is 1, Medium is 2, and Large is 3. This order works for sizes, but not every kind of category.

**Summary:** Cleaning fixes empty, repeated, or messy data. Encoding changes categories into a form that is easier to use in calculations.
