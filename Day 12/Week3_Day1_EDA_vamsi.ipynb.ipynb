# 4. Which product, category, or item appears most often?

```python
subcategory_counts = df["Sub-Category"].value_counts()

print("Most frequent sub-category:", subcategory_counts.idxmax())
print("Number of records:", subcategory_counts.max())

display(subcategory_counts.head(10))
```

Explanation:
- value_counts() counts the records for each sub-category.
- idxmax() returns the sub-category with the highest count.
- max() returns the highest count.

Finding: The sub-category shown by the output appears most frequently in the dataset.

# 5. What is the total sales revenue?

```python
total_sales = df["Sales"].sum()

print("Total Sales:", round(total_sales, 2))
```

Explanation: The sum() function adds the sales amounts across all rows.

Finding: The total recorded sales revenue is the amount displayed in the output.

# 6. What is the average order value?

```python
order_totals = df.groupby("Order ID")["Sales"].sum()

average_order_value = order_totals.mean()

print("Number of unique orders:", df["Order ID"].nunique())
print("Average Order Value:", round(average_order_value, 2))
```

Explanation:
- groupby("Order ID") groups rows belonging to the same order.
- sum() calculates the total sales for each order.
- mean() calculates the average order value.
- nunique() counts unique order IDs.

Finding: The average order value represents the average total sales amount per unique order.

# 7. Which record has the highest sales value?

```python
highest_sales_index = df["Sales"].idxmax()
highest_sales_record = df.loc[highest_sales_index]

display(
    highest_sales_record[
        ["Order ID", "Order Date", "Product Name",
         "Category", "Region", "Sales", "Profit"]
    ]
)
```

Explanation:
- idxmax() finds the index of the row with the highest sales value.
- loc[] retrieves that record.

Finding: The displayed record identifies the order and product with the highest recorded sales value.

# 8. Which record has the lowest sales value?

```python
lowest_sales_index = df["Sales"].idxmin()
lowest_sales_record = df.loc[lowest_sales_index]

display(
    lowest_sales_record[
        ["Order ID", "Order Date", "Product Name",
         "Category", "Region", "Sales", "Profit"]
    ]
)
```

Explanation:
- idxmin() finds the index of the row with the lowest sales value.
- loc[] retrieves that record.

Finding: The displayed record identifies the order and product with the lowest recorded sales value.

# 9. Which category has the highest total sales?

```python
category_sales = (
    df.groupby("Category")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

display(category_sales)

print("Category with highest sales:", category_sales.idxmax())
print("Highest category sales:", round(category_sales.max(), 2))
```

Explanation:
- groupby("Category") groups the data by product category.
- sum() calculates total sales for each category.
- sort_values() arranges the results from highest to lowest.

Finding: The category identified by the output generates the highest total sales.
