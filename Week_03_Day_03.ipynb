```python
# line plot - follow one student's scores
import matplotlib.pyplot as plt
quizzes = ['Quiz 1', 'Quiz 2', 'Quiz 3', 'Quiz 4']
scores = [58, 66, 72, 81]
plt.plot(quizzes, scores, marker='o')
plt.title("Vamsi's Quiz Scores")
plt.xlabel('Quiz')
plt.ylabel('Score')
plt.show()
```

```python
# bar chart - compare student scores
import matplotlib.pyplot as plt
students = ['Vamsi', 'Abhi', 'Kiran']
scores = [81, 74, 89]
plt.bar(students, scores)
plt.title('Student Score Comparison')
plt.xlabel('Student')
plt.ylabel('Score')
plt.show()
```

```python
# histogram - count scores in score ranges
import matplotlib.pyplot as plt
scores = [52, 60, 64, 69, 73, 77, 82, 86, 91, 96]
plt.hist(scores, bins=5, edgecolor='black')
plt.title('Student Score Ranges')
plt.xlabel('Score')
plt.ylabel('Number of Students')
plt.show()
```

```python
# scatter plot - compare study time and test scores
import matplotlib.pyplot as plt
study_hours = [1, 2, 3, 4, 5]
test_scores = [55, 61, 70, 78, 88]
plt.scatter(study_hours, test_scores)
plt.title('Study Hours and Test Scores')
plt.xlabel('Study hours')
plt.ylabel('Test score')
plt.show()
```

```python
# labels and legend - compare two students over time
import matplotlib.pyplot as plt
terms = [1, 2, 3]
vamsi_scores = [65, 72, 81]
abhi_scores = [70, 74, 79]
plt.plot(terms, vamsi_scores, marker='o', label='Vamsi')
plt.plot(terms, abhi_scores, marker='o', label='Abhi')
plt.title('Scores by Term')
plt.xlabel('Term')
plt.ylabel('Score')
plt.legend()
plt.show()
```

```python
# check whether scores went up overall
scores = [58, 66, 72, 81]
if scores[-1] > scores[0]:
    print('The scores went up overall')
else:
    print('The scores did not go up overall')
```

## Simple Ideas and Results

**1. What is data visualization?**  
It means showing information with charts. Charts can make patterns easier to see.

**2. What is Matplotlib?**  
Matplotlib is a Python tool for drawing charts.

**3. What is a line plot?**  
A line plot joins points to show change over time. In the first chart, Vamsi's score rises across four quizzes.

**4. What is a bar chart?**  
A bar chart compares groups. The chart compares the scores of Vamsi, Abhi, and Kiran.

**5. What is a histogram?**  
A histogram groups numbers into ranges and counts them. A taller bar means more scores fall in that range.

**6. What is a scatter plot?**  
A scatter plot uses dots to compare two number lists. Here, each dot shows study hours and a test score.

**7. What are labels and a legend?**  
The title explains the chart. Axis labels explain each direction. The legend tells us which line belongs to Vamsi and Abhi.

**8. What is a trend?**  
A trend is the general direction of values. The last score (81) is higher than the first (58), so the scores went up overall.

**Summary:** Line plots show change, bars compare groups, histograms count ranges, and scatter plots compare two number lists.
